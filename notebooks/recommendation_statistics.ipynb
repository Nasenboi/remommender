# Analyze the statistics of the recommendation process

```python
# imports
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
statistics_path = "/path/to/recommendation_statistics.csv"
data = pd.read_csv(statistics_path)
```

```python
data.columns
```

```python
SONG_SWITCH_THRESHOLD = 0.5
```

```python
data['timestamp'] = pd.to_datetime(data['timestamp'])

# Subtract the first timestamp from all timestamps to get relative time
relative_time = data['timestamp'] - data['timestamp'].iloc[0]
# Format as HH:MM:SS
time_labels = relative_time.dt.components['hours'].astype(str).str.zfill(2) + ':' + \
              relative_time.dt.components['minutes'].astype(str).str.zfill(2) + ':' + \
              relative_time.dt.components['seconds'].astype(str).str.zfill(2)

plt.figure(figsize=(14, 6))
plt.plot(time_labels, data['valence'], color='blue', label='Valence')
plt.plot(time_labels, data['arousal'], color='red', label='Arousal')
plt.plot(time_labels, data['song_valence'], color='blue', linestyle=':', label='Song Valence')
plt.plot(time_labels, data['song_arousal'], color='red', linestyle=':', label='Song Arousal')

# Add vertical gray lines where switch_probability > SONG_SWITCH_THRESHOLD
switch_indices = data['switch_probability'] > SONG_SWITCH_THRESHOLD
for x in np.array(time_labels)[switch_indices]:
    plt.axvline(x=x, color='gray', linestyle='--', alpha=0.5)

plt.xlabel('Time since start (HH:MM:SS)')
plt.ylabel('Value')
plt.title('Valence (blue) and Arousal (red) over Time')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
